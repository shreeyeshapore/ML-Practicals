# Experiment: K-Means Clustering

**Aim:** To apply K-Means clustering and evaluate the clusters using the Elbow Method and Silhouette Score.

**Libraries:** NumPy, Matplotlib and Scikit-learn

## 1. Import libraries and create data

```python
## 1. Import libraries and create data

import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Create simple sample data with 4 groups
X, y = make_blobs(n_samples=250, centers=4, random_state=10)

print("Dataset created successfully")
print("Number of samples:", len(X))
```

## 2. Display the dataset

```python
## 2. Display the dataset

plt.scatter(X[:, 0], X[:, 1], c='green')
plt.xlabel("Feature X")
plt.ylabel("Feature Y")
plt.title("Dataset before Applying K-Means")
plt.show()
```

## 3. Find the number of clusters using the Elbow Method

```python
## 3. Find the number of clusters using the Elbow Method

# Calculate WCSS for different numbers of clusters
wcss = []

for k in range(1, 9):
    model = KMeans(n_clusters=k, random_state=10, n_init=15)
    model.fit(X)
    wcss.append(model.inertia_)

# Plot the Elbow graph
plt.plot(range(1, 9), wcss, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

**Observation:** WCSS decreases as the number of clusters increases. The optimal number of clusters is identified at the point where the decrease in WCSS becomes less significant, known as the elbow point. In this dataset, the elbow is expected around K = 4, indicating four natural clusters in the data.

## 4. Apply K-Means with 3 clusters

```python
## 4. Apply K-Means with 4 clusters

# Create and train the K-Means model
model = KMeans(n_clusters=4, random_state=10, n_init=15)
labels = model.fit_predict(X)

# Display cluster centres
print("Cluster Centres:")
print(model.cluster_centers_)
```

## 5. Visualize the clusters

```python
## 5. Visualize the clusters

# Plot data points with their cluster labels
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis")

# Plot cluster centres
plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    color="red",
    marker="D",
    s=180,
    label="Cluster Centres"
)

plt.xlabel("Feature X")
plt.ylabel("Feature Y")
plt.title("K-Means Cluster Visualization")
plt.legend()
plt.show()
```

**Observation:** K-Means assigns each data point to the nearest cluster centre. Points belonging to the same cluster are displayed in the same colour, and the cluster centres indicate the representative position of each cluster.

## 6. Evaluate using Silhouette Score

```python
## 6. Evaluate using Silhouette Score

# Calculate Silhouette Score
score = silhouette_score(X, labels)

print("Silhouette Score:", round(score, 4))
```

**Observation:** The Silhouette Score ranges from -1 to +1. A score closer to +1 indicates well-separated clusters, while a score near 0 or negative suggests overlapping or poorly assigned clusters.

## Result

K-Means clustering was successfully applied to the dataset. The Elbow Method was used to determine a suitable number of clusters, and the Silhouette Score was calculated to evaluate the quality of clustering. This helped assess how effectively the data points were grouped.
