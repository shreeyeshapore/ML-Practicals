# Experiment: Toy Classification Using Decision Boundaries (Manual Coding)


## Aim
To implement a simple toy classification problem using a manually defined decision boundary and visualize the classification results.

## Step 1: Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Create a Toy Dataset

The dataset has two input features (`X1`, `X2`) and two classes:
- Class 0
- Class 1

```python
## Step 2: Create a Toy Dataset

# The dataset has two input features (X1, X2) and two classes:
# Class 0
# Class 1

X = np.array([
    [2, 1], [3, 1], [2, 2], [3, 2],
    [6, 5], [7, 5], [6, 6], [7, 6]
])

y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

print("Input data:")
print(X)

print("Actual classes:")
print(y)
```

## Step 3: Define a Decision Boundary Manually

Use the line:

**X1 + X2 = 6**

- If `X1 + X2 >= 6`, predict Class 1.
- Otherwise, predict Class 0.

```python
## Step 3: Define a Decision Boundary Manually

# Use the line:
# X1 + X2 = 8

# If X1 + X2 >= 8, predict Class 1.
# Otherwise, predict Class 0.

y_pred = []

for point in X:
    if point[0] + point[1] >= 8:
        y_pred.append(1)
    else:
        y_pred.append(0)

print("Predicted classes:")
print(y_pred)
```

## Step 4: Calculate Accuracy

```python
## Step 4: Calculate Accuracy

correct = 0

for i in range(len(y)):
    if y[i] == y_pred[i]:
        correct += 1

accuracy = (correct / len(y)) * 100

print("Number of Correct Predictions:", correct)
print("Classification Accuracy:", accuracy, "%")
```

## Step 5: Visualize the Dataset and Decision Boundary

```python
## Step 5: Visualize the Dataset and Decision Boundary

plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors="blue")

x_line = np.array([0, 8])
y_line = 8 - x_line

plt.plot(x_line, y_line, "g--", label="X1 + X2 = 8")

plt.xlabel("Input Feature X1")
plt.ylabel("Input Feature X2")
plt.title("Manual Decision Boundary for Toy Classification")
plt.legend()
plt.grid()
plt.show()
```

## Step 6: Test a New Data Point

Enter a new point and classify it using the same manually defined boundary.

```python
## Step 6: Test a New Data Point

# Enter a new point and classify it using the same manually defined boundary.

new_point = [4, 5]

if new_point[0] + new_point[1] >= 8:
    print("New point belongs to Class 1")
else:
    print("New point belongs to Class 0")
```

## Observations

1. The toy dataset contains two features and two classes.
2. The decision boundary is defined manually using the equation X1 + X2 = 8.
3. Points with a sum greater than or equal to 8 are classified as Class 1.
4. Points with a sum less than 8 are classified as Class 0.
5. Accuracy is calculated by comparing actual and predicted class labels.
6. A new data point can also be classified using the same decision rule.

## Result

A toy binary classification problem was successfully implemented using manual decision-boundary logic. The class predictions, accuracy, and decision-boundary plot were obtained. The model correctly classified the data based on the defined decision rule.
